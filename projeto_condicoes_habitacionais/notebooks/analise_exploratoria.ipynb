```
from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path('..')
df = pd.read_csv(ROOT / 'data/processed/problemas_domiciliares.csv')
df.head()
```

## Verificações de qualidade

```
print('Dimensões:', df.shape)
print('Nulos por coluna:\n', df.isna().sum())
print('Capitais duplicadas:', df['capital'].duplicated().sum())
indicadores = [c for c in df.columns if c != 'capital']
assert df[indicadores].apply(lambda s: s.between(0, 100).all()).all()
```

## Estatística descritiva

```
descricao = df[indicadores].describe().T
descricao
```

## Médias por indicador e por capital

```
medias_problemas = df[indicadores].mean().sort_values(ascending=False)
medias_capitais = df.assign(media_simples=df[indicadores].mean(axis=1)).sort_values('media_simples', ascending=False)
medias_problemas, medias_capitais[['capital','media_simples']]
```

## Correlação exploratória

As correlações são calculadas sobre somente 12 observações agregadas (capitais). Devem ser interpretadas com cautela e não indicam causalidade.

```
correlacoes = df[indicadores].corr()
correlacoes.round(2)
```
